```python
# Проверяем, виден ли GPU
import torch

print("PyTorch версия:", torch.__version__)
print("CUDA доступна:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
    print("Память GPU:", round(torch.cuda.get_device_properties(0).total_memory / 1e9, 1), "GB")
else:
    print("⚠️ GPU не виден! Runtime → Change runtime type → T4 GPU")
```

```python
# Создаём тензор — это "массив чисел", базовый тип в нейросетях
a = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
b = torch.tensor([[5.0, 6.0], [7.0, 8.0]])

print("a на CPU:", a.device)

# Переносим на GPU — одна строчка
a_gpu = a.to("cuda")
b_gpu = b.to("cuda")

print("a на GPU:", a_gpu.device)

# Матричное умножение на GPU
c = a_gpu @ b_gpu
print("Результат a @ b:\n", c.cpu())  # вернули на CPU для печати
```
